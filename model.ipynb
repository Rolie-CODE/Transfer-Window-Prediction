```python
data1 = pd.read_csv('data/data1.csv')
```

```python
print(data1.head())
```

```python
data1.isnull().sum()
```

```python
target = data1['market_value_in_eur']
new_set = data1.drop(['market_value_in_eur', 'market_value_date', 'appearance_id', 'date', 'competition_id' , 'player_name'], axis = 1)
```

```python
print(target)
print(new_set)
```

```python
X_train, X_test, y_train, y_test = train_test_split(new_set, target, test_size=0.1, random_state=42)
```

```python
print(X_train.dtypes)
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
prediction = model.predict(X_test)
print(prediction)
```

```python
from sklearn.metrics import mean_absolute_error
```

```python
mae  = mean_absolute_error(y_test, prediction)
```

```python
print(f"Mean Absolute Error (MAE): €{mae:,.2f}")
```

```python
model_filename = 'transfer_window_model.pkl'
with open(model_filename, 'wb') as file:
    pickle.dump(model, file)

print(f"Model successfully saved to {model_filename}")
```
